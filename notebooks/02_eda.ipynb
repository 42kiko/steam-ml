```python
from pathlib import Path

import pandas as pd
import plotly.express as px

# Assume this notebook lives in `notebooks/` under the project root.
PROJECT_ROOT = Path('..').resolve()
DATA_DIR = PROJECT_ROOT / 'data'
WEB_PARQUET = DATA_DIR / 'bronze' / 'steam_web' / 'app_list.parquet'
STORE_PARQUET = DATA_DIR / 'bronze' / 'steam_store' / 'app_details.parquet'
SPY_PARQUET = DATA_DIR / 'bronze' / 'steam_spy' / 'all.parquet'

WEB_PARQUET, STORE_PARQUET, SPY_PARQUET
```

```python
df_web = pd.read_parquet(WEB_PARQUET)
df_store = pd.read_parquet(STORE_PARQUET)
df_spy = pd.read_parquet(SPY_PARQUET)
```

```python
df_web.info(), df_store.info(), df_spy.info()
```

```python
df_web["last_modified"] = pd.to_datetime(
    df_web["last_modified"], unit="s", utc=True
)
```

```python
fig = px.histogram(
    df_spy,
    x="owners",
    nbins=60,
    log_y=True,
    title="Steam Games Long-Tail Distribution (Owners)",
)

fig.update_xaxes(title="Owners (Range Index)")
fig.update_yaxes(title="Number of Games (log scale)")

fig.show()
```

```python
days_since_update = (
    pd.Timestamp.utcnow()
    - df_web["last_modified"]
).dt.days

fig = px.histogram(
    days_since_update,
    nbins=60,
    title="Days Since Last Update (All Steam Apps)",
)

fig.update_xaxes(title="Days since last modification")
fig.update_yaxes(title="Number of apps")

fig.show()
```

```python
merged = df_web.merge(
    df_spy[["appid", "ccu"]],
    on="appid",
    how="inner",
)

merged["days_since_update"] = (
    pd.Timestamp.utcnow() - merged["last_modified"]
).dt.days

fig = px.scatter(
    merged,
    x="days_since_update",
    y="ccu",
    log_y=True,
    opacity=0.4,
    title="Update Recency vs CCU",
)

fig.update_xaxes(title="Days since last update")
fig.update_yaxes(title="CCU (log)")

fig.show()
```

```python
df_web["year_month"] = df_web["last_modified"].dt.to_period("M").astype(str)

apps_over_time = (
    df_web.groupby("year_month")
    .size()
    .reset_index(name="num_apps")
)

fig = px.line(
    apps_over_time,
    x="year_month",
    y="num_apps",
    title="Steam App Catalog Growth Over Time",
)

fig.update_xaxes(title="Year–Month")
fig.update_yaxes(title="Number of Apps")

fig.show()
```

```python
df_web["days_since_update"] = (
    pd.Timestamp.utcnow() - df_web["last_modified"]
).dt.days

fig = px.histogram(
    df_web,
    x="days_since_update",
    nbins=80,
    title="Days Since Last Update (All Steam Apps)",
)

fig.add_vline(x=30, line_dash="dash", annotation_text="< 30 days")
fig.add_vline(x=180, line_dash="dash", annotation_text="30–180 days")
fig.add_vline(x=365, line_dash="dash", annotation_text="> 1 year")

fig.update_xaxes(title="Days since last update")
fig.update_yaxes(title="Number of apps")

fig.show()
```

```python
df_web["name_length"] = df_web["name"].str.len()
df_web["word_count"] = df_web["name"].str.split().str.len()

fig = px.histogram(
    df_web,
    x="name_length",
    nbins=40,
    title="Distribution of Game Name Lengths",
)

fig.show()
```

```python
fig = px.histogram(
    df_web,
    x="word_count",
    nbins=20,
    title="Distribution of Words per Game Title",
)

fig.show()
```

```python
fig = px.histogram(
    df_spy,
    x="owners",
    nbins=60,
    log_y=True,
    title="Steam Long-Tail Distribution (Owners)",
)

fig.update_xaxes(title="Owners")
fig.update_yaxes(title="Number of Games (log scale)")

fig.show()
```

```python
fig = px.scatter(
    df_spy,
    x="owners",
    y="average_forever",
    log_x=True,
    log_y=True,
    opacity=0.5,
    title="Average Playtime vs Owners",
)

fig.update_xaxes(title="Owners (log)")
fig.update_yaxes(title="Average Playtime (log)")

fig.show()
```
