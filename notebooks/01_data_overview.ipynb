```python
from pathlib import Path

import pandas as pd
import plotly.express as px

# Assume this notebook lives in `notebooks/` under the project root.
PROJECT_ROOT = Path('..').resolve()
DATA_DIR = PROJECT_ROOT / 'data'
WEB_PARQUET = DATA_DIR / 'bronze' / 'steam_web' / 'app_list.parquet'
STORE_PARQUET = DATA_DIR / 'bronze' / 'steam_store' / 'app_details.parquet'
SPY_PARQUET = DATA_DIR / 'bronze' / 'steam_spy' / 'all.parquet'

WEB_PARQUET, STORE_PARQUET, SPY_PARQUET
```

```python
# plotty theme
import plotly.io as pio

pio.templates.default = "plotly_dark"
```

```python
web_df = pd.read_parquet(WEB_PARQUET)
store_df = pd.read_parquet(STORE_PARQUET)
spy_df = pd.read_parquet(SPY_PARQUET)
```

```python
web_df.info(), store_df.info(), spy_df.info()
```

```python
web_df["last_modified"] = pd.to_datetime(
    web_df["last_modified"], unit="s", utc=True
)
```

```python
web_df["year_month"] = web_df["last_modified"].dt.to_period("M").astype(str)

apps_over_time = (
    web_df.groupby("year_month")
    .size()
    .reset_index(name="num_apps")
)

fig = px.line(
    apps_over_time,
    x="year_month",
    y="num_apps",
    title="Steam App Catalog Growth Over Time",
)

fig.update_xaxes(title="Year-Month")
fig.update_yaxes(title="Number of Apps")

fig.show()
```

```python
web_df["days_since_update"] = (
    pd.Timestamp.utcnow() - web_df["last_modified"]
).dt.days

fig = px.histogram(
    web_df,
    x="days_since_update",
    nbins=80,
    title="Days Since Last Update (All Steam Apps)",
)

fig.add_vline(x=30, line_dash="dash", annotation_text="< 30 days")
#fig.add_vline(x=180, line_dash="dash", annotation_text="30-180 days")
fig.add_vline(x=365, line_dash="dash", annotation_text="> 1 year")

fig.update_xaxes(title="Days since last update")
fig.update_yaxes(title="Number of apps")

fig.show()
```

```python
coverage = pd.DataFrame({
    "App List": web_df["appid"].notna().mean(),
    "SteamSpy": web_df["appid"].isin(spy_df["appid"]).mean(),
    "Store details": web_df["appid"].isin(store_df["appid"]).mean(),
}, index=["coverage"]).T.reset_index()

coverage.columns = ["dataset", "coverage"]

fig = px.bar(
    coverage,
    x="dataset",
    y="coverage",
    title="Dataset Coverage by Source",
)

fig.update_yaxes(range=[0, 1])

fig.show()
```

```python
import ast


def parse_genres(value):
    if pd.isna(value):
        return []
    if isinstance(value, list):
        return value
    if isinstance(value, str):
        try:
            return ast.literal_eval(value)
        except Exception:
            return []
    return []

store_df["genres_parsed"] = store_df["genres"].apply(parse_genres)

store_df["genres_parsed"].head()
```

```python
genres_exploded = (
    store_df[["steam_appid", "genres_parsed"]]
    .explode("genres_parsed")
    .dropna()
)

genres_exploded["genre"] = genres_exploded["genres_parsed"].apply(
    lambda g: g.get("description") if isinstance(g, dict) else None
)


genres_exploded = genres_exploded.dropna(subset=["genre"])

genre_success = (
    genres_exploded
    .merge(
        spy_df[["appid", "owners_num"]],
        left_on="steam_appid",
        right_on="appid",
        how="left",
    )
    .groupby("genre", as_index=False)["owners_num"]
    .quantile(0.75)
    .sort_values("owners_num", ascending=False)
    .head(15)
)

fig = px.bar(
    genre_success,
    x="owners_num",
    y="genre",
    orientation="h",
    title="Median Owners per Genre (Partial Coverage)",
)

fig.show()
```

```python
success_threshold = 100_000

genre_success_rate = (
    genres_exploded
    .merge(spy_df[["appid", "owners_num"]], left_on="steam_appid", right_on="appid")
    .assign(success=lambda df: df["owners_num"] >= success_threshold)
    .groupby("genre")["success"]
    .mean()
    .reset_index()
    .sort_values("success", ascending=False)
)

fig = px.box(
    genres_exploded
    .merge(spy_df[["appid", "owners_num"]], left_on="steam_appid", right_on="appid"),
    x="owners_num",
    y="genre",
    log_x=True,
    title="Owners Distribution per Genre",
)

fig.show()
```

```python
def parse_required_age(x):
    if pd.isna(x):
        return 0
    if isinstance(x, str):
        x = x.strip()
        if x.endswith("+"):
            return int(x.replace("+", ""))
        if x.isdigit():
            return int(x)
        return 0
    return int(x)

store_df["required_age_num"] = store_df["required_age"].apply(parse_required_age)
```

```python
age_counts = (
    store_df["required_age_num"]
    .value_counts()
    .sort_index()
    .reset_index(name="count")
)

fig = px.bar(
    age_counts,
    x="required_age_num",
    y="count",
    title="Required Age Distribution on Steam",
)

fig.update_xaxes(title="Required Age")
fig.update_yaxes(title="Number of Games")

fig.show()

```

```python
fig = px.bar(
    age_counts,
    x="required_age_num",
    y="count",
    log_y=True,
    title="Required Age Distribution on Steam (log scale)",
)

fig.show()
```
